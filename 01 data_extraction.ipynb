```python
# Clean and normalize both the employee summary and the job description.
import re
# python -m spacy download en_core_web_md
import spacy
nlp = spacy.load("en_core_web_md")
def preprocess(text):
    text = text.lower()
    text = re.sub(r'[^\w\s]', '', text)
    return text

candidate_text = "I am a data scientist with 3 years of experience in Python, SQL, and machine learning. I deployed models using AWS."

result = preprocess(candidate_text)
print(result)
```

```python
# Option B: Using Sentence-BERT (more accurate for longer texts)
# 
from sentence_transformers import SentenceTransformer, util
model = SentenceTransformer('all-MiniLM-L6-v2')

def get_bert_similarity(text1, text2):
    embeddings = model.encode([text1, text2], convert_to_tensor=True)
    return float(util.pytorch_cos_sim(embeddings[0], embeddings[1]))
```

3. Optional: Include Keyword-Based Boost

```python
#Create a list of must-have keywords/skills from the job description. Boost score based on keyword overlap with employee summary.

def keyword_boost(text1, text2, keywords):
    count = sum(1 for kw in keywords if kw.lower() in text1.lower() and kw.lower() in text2.lower())
    return count / len(keywords)  # normalized boost factor
```

```python
# Combine Semantic + Keyword Score (Optional Hybrid Model)
def final_similarity_score(summary, job_desc, keywords=[]):
    sim_score = get_bert_similarity(summary, job_desc)
    boost = keyword_boost(summary, job_desc, keywords)
    return round(0.8 * sim_score + 0.2 * boost, 3)
```

```python
from sentence_transformers import SentenceTransformer, util
from transformers import pipeline
```

```python

# Load models
sbert_model = SentenceTransformer('all-MiniLM-L6-v2')
nli_model = pipeline("text-classification", model="roberta-large-mnli")

def combined_similarity(text1, text2, sbert_threshold=0.5):
    # Step 1: SBERT similarity
    embeddings = sbert_model.encode([text1, text2], convert_to_tensor=True)
    sim_score = float(util.pytorch_cos_sim(embeddings[0], embeddings[1]))

    # If similarity is low, no need to do NLI
    if sim_score < sbert_threshold:
        return {"similarity": sim_score, "relation": "unrelated"}

    # Step 2: Run NLI for sentence relationship
    nli_input = f"{text1} </s></s> {text2}"
    nli_result = nli_model(nli_input)[0]

    return {
        "similarity": sim_score,
        "relation": nli_result['label'],  # ENT, CON, NEUTRAL
        "confidence": nli_result['score']
    }
```

```python
from sentence_transformers import SentenceTransformer, util
from transformers import pipeline
import spacy
# python -m spacy download en_core_web_sm

# Load models
sbert_model = SentenceTransformer("all-MiniLM-L6-v2")
nli_model = pipeline("text-classification", model="roberta-large-mnli")
nlp = spacy.load("en_core_web_sm")

# Simple skill extractor using SpaCy
def extract_skills(text, keywords):
    doc = nlp(text.lower())
    found = set()
    for token in doc:
        if token.text in keywords:
            found.add(token.text)
    return found

# Match score calculator
def get_match_score(candidate_text, job_text, known_skills):
    # 1. Skill overlap
    cand_skills = extract_skills(candidate_text, known_skills)
    job_skills = extract_skills(job_text, known_skills)
    overlap = cand_skills.intersection(job_skills)
    skill_score = len(overlap) / len(job_skills) if job_skills else 0

    # 2. Semantic similarity
    embeddings = sbert_model.encode([candidate_text, job_text], convert_to_tensor=True)
    sim_score = float(util.pytorch_cos_sim(embeddings[0], embeddings[1]))

    # 3. NLI relationship
    relation = nli_model(f"{candidate_text} </s></s> {job_text}")[0]
    nli_bonus = 1.0 if relation['label'] == 'ENTAILMENT' else 0.5 if relation['label'] == 'NEUTRAL' else 0

    # Final weighted score (out of 100)
    final_score = (skill_score * 40) + (sim_score * 40) + (nli_bonus * 20)
    return round(final_score, 2), {
        "skills_matched": list(overlap),
        "semantic_score": round(sim_score, 2),
        "nli_relation": relation['label']
    }
```

Sample 1

```python
candidate_text = "I am a data scientist with 3 years of experience in Python, SQL, and machine learning. I deployed models using AWS."
job_text = "Looking for a machine learning engineer with experience in Python, SQL, AWS, and model deployment."

known_skills = {"python", "sql", "aws", "machine learning", "model deployment", "pandas", "scikit-learn"}

score, details = get_match_score(candidate_text, job_text, known_skills)

print("Match Score:", score)
print("Details:", details)
```
