```python
import numpy as np
import pandas as pd
import json
```

```python
# Load the JSON file
with open('linkedIn_data.json', 'r') as file:
    data = json.load(file)

# Flatten and normalize nested JSON structures
df_profiles = pd.json_normalize(data)

# Show the first few rows
print(df_profiles.head())
df_profiles
```

```python
# Data Cleaning
df_profiles['location'] = df_profiles.apply(lambda row: {'location.postalCode': row['location.postalCode'], 'location.geographicArea': row['location.geographicArea']}, axis=1)

df_profiles['posts_summary']=df_profiles.apply(lambda row: {'total_posts':row['total_posts'],'total_likes':row['posts_summary.total_likes'],'total_comments':row['posts_summary.total_comments'],'total_shares':row['posts_summary.total_shares']},axis=1)
df_profiles['fullName']= df_profiles['firstName.localized.en_US'] + ' ' + df_profiles['lastName.localized.en_US']
df_profiles.columns
```

```python
# Making a new clean dataset with well formated and relevant information
profiles_data=df_profiles[['id','firstName.localized.en_US','lastName.localized.en_US','fullName','publicProfileUrl','headline.localized.en_US','industry','summary.localized.en_US','location','education.elements','certifications','positions.elements','skills.elements','followers','posts','posts_summary']]

profiles_data
```
